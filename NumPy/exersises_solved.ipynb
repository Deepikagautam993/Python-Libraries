**Valid Sudoku**

Check whether every row, column, and 3×3 subgrid contains the digits 1–9 exactly once.

```python
s = np.array([
    [5, 3, 4, 6, 7, 8, 9, 1, 2],
    [6, 7, 2, 1, 9, 5, 3, 4, 8],
    [1, 9, 8, 3, 4, 2, 5, 6, 7],
    
    [8, 5, 9, 7, 6, 1, 4, 2, 3],
    [4, 2, 6, 8, 5, 3, 7, 9, 1],
    [7, 1, 3, 9, 2, 4, 8, 5, 6],
    
    [9, 6, 1, 5, 3, 7, 2, 8, 4],
    [2, 8, 7, 4, 1, 9, 6, 3, 5],
    [3, 4, 5, 2, 8, 6, 1, 7, 9]
])


# Check whether the Sudoku board is valid.
# Each row, column, and 3x3 subgrid must contain digits 1 through 9 exactly once.
expected = set(range(1, 10))

rows_valid = all(set(row) == expected for row in s)
columns_valid = all(set(s[:, col]) == expected for col in range(9))
boxes_valid = all(
    set(s[r:r+3, c:c+3].flatten()) == expected
    for r in range(0, 9, 3)
    for c in range(0, 9, 3)
)

is_valid = rows_valid and columns_valid and boxes_valid
print("Valid Sudoku:", is_valid)
```

**General Questions**

Use NumPy indexing, filtering, aggregation, and array operations to solve the following exercises.

```python
import numpy as np

# Columns: [Age, Math Marks, Science Marks]
data = np.array([
    [18, 85, 78],   # Student 1
    [19, 92, 88],   # Student 2
    [17, 76, 95],   # Student 3
    [18, 65, 70],   # Student 4
    [20, 90, 85]    # Student 5
])
```

```python
# Get the shape of the matrix.
print(data.shape)
```

```python
# Find the average age of students.
print(data[:, 0].mean())
```

```python
# Extract Math marks of all students.
print(data[:, 1])
```

```python
# Find the highest Science mark.
print(data[:, 2].max())
```

```python
# Get details of the student who scored more than 90 in Math.
print(data[data[:, 1] > 90])
```

```python
# Increase Math marks of all students by 5.
data_updated = data.copy()
data_updated[:, 1] += 5
print(data_updated)
```

```python
# Find how many students are younger than 19.
print(np.sum(data[:, 0] < 19))
```

```python
# Calculate the average marks in each subject (column-wise mean).
print(data[:, 1:].mean(axis=0))
```

```python
# Get data of students who scored at least 80 in both subjects.
print(data[(data[:, 1] >= 80) & (data[:, 2] >= 80)])
```

```python
# Replace all Science marks < 75 with 0.
data_updated = data.copy()
data_updated[data_updated[:, 2] < 75, 2] = 0
print(data_updated)
```
